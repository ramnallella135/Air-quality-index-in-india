```python

import pandas as pd
import numpy as np
import os

# If your dataset is already loaded as df, skip this line.
# Otherwise, use the path of your uploaded CSV file:
# df = pd.read_csv(file_path)

# Check that the dataset is loaded
print("Original dataset shape:", df.shape)
display(df.head())
```

```python

# Check column names and data types
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python

# Create a copy so the original dataset remains unchanged
df_cleaned = df.copy()

# 1. Remove extra spaces from column names
df_cleaned.columns = (
    df_cleaned.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# 2. Remove leading and trailing spaces from text columns
for col in df_cleaned.select_dtypes(include=["object", "string"]).columns:
    df_cleaned[col] = df_cleaned[col].str.strip()

# 3. Replace empty strings and common missing-value markers with NaN
df_cleaned.replace(
    r"^\s*$",
    np.nan,
    regex=True,
    inplace=True
)

df_cleaned.replace(
    ["NA", "N/A", "na", "n/a", "NULL", "null", "?"],
    np.nan,
    inplace=True
)

# 4. Remove duplicate rows
df_cleaned.drop_duplicates(inplace=True)

# 5. Remove columns that contain no data at all
df_cleaned.dropna(axis=1, how="all", inplace=True)

# 6. Fill missing values
for col in df_cleaned.columns:

    if pd.api.types.is_numeric_dtype(df_cleaned[col]):
        # Fill numeric missing values with the median
        if df_cleaned[col].notna().any():
            df_cleaned[col] = df_cleaned[col].fillna(
                df_cleaned[col].median()
            )

    else:
        # Fill categorical/text missing values with the mode
        mode_values = df_cleaned[col].mode(dropna=True)

        if not mode_values.empty:
            df_cleaned[col] = df_cleaned[col].fillna(
                mode_values.iloc[0]
            )
        else:
            df_cleaned[col] = df_cleaned[col].fillna("Unknown")

print("Dataset cleaning completed!")
print("Original shape:", df.shape)
print("Cleaned shape:", df_cleaned.shape)

display(df_cleaned.head())
```

```python

print("Missing values after cleaning:")
print(df_cleaned.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df_cleaned.duplicated().sum())

print("\nCleaned dataset information:")
df_cleaned.info()
```

```python

# Save the cleaned dataset
output_file = os.path.abspath("cleaned_dataset.csv")

df_cleaned.to_csv(output_file, index=False, encoding="utf-8-sig")

print("Cleaned dataset saved successfully!")
print("File location:", output_file)
```
