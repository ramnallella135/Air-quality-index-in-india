```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\RAM\Downloads\cleaned_dataset.csv")
print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
print("Number of cities:", df["city"].nunique())
```

```python
print(df["city"].unique())
```

```python
city_counts = df["city"].value_counts()

print(city_counts)
```

```python
average_aqi = df["aqi"].mean()

print("Average AQI:", round(average_aqi, 2))
```

```python
print("Minimum AQI:", df["aqi"].min())
print("Maximum AQI:", df["aqi"].max())
```

```python
print("AQI Mean:", round(df["aqi"].mean(), 2))
print("AQI Median:", df["aqi"].median())
print("AQI Standard Deviation:", round(df["aqi"].std(), 2))
```

```python
aqi_categories = df["aqi_bucket"].value_counts()

print(aqi_categories)
```

```python
aqi_percentage = df["aqi_bucket"].value_counts(normalize=True) * 100

print(aqi_percentage.round(2))
```

```python
city_aqi = df.groupby("city")["aqi"].mean().sort_values(ascending=False)

print(city_aqi.round(2))
```

```python
highest_aqi_city = city_aqi.idxmax()
highest_aqi_value = city_aqi.max()

print("City with highest average AQI:", highest_aqi_city)
print("Average AQI:", round(highest_aqi_value, 2))
```

```python
lowest_aqi_city = city_aqi.idxmin()
lowest_aqi_value = city_aqi.min()

print("City with lowest average AQI:", lowest_aqi_city)
print("Average AQI:", round(lowest_aqi_value, 2))
```

```python
pollutants = [
    "pm2.5", "pm10", "no", "no2", "nox",
    "nh3", "co", "so2", "o3",
    "benzene", "toluene", "xylene"
]

pollutant_average = df[pollutants].mean().sort_values(ascending=False)

print(pollutant_average.round(2))
```

```python
print(df[pollutants].min().round(2))
```

```python
print(df[pollutants].max().round(2))
```

```python
pm25_city = df.groupby("city")["pm2.5"].mean().sort_values(ascending=False)

print(pm25_city.round(2))
```

```python
pm10_city = df.groupby("city")["pm10"].mean().sort_values(ascending=False)

print(pm10_city.round(2))
```

```python
city = pm25_city.idxmax()
value = pm25_city.max()

print("Highest average PM2.5:", city)
print("Value:", round(value, 2))
```

```python
city = pm10_city.idxmax()
value = pm10_city.max()

print("Highest average PM10:", city)
print("Value:", round(value, 2))
```

```python
aqi_correlation = df[pollutants + ["aqi"]].corr()["aqi"].sort_values(ascending=False)

print(aqi_correlation.round(3))
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")

df["year"] = df["date"].dt.year

print(df["year"].unique())
```

```python
yearly_aqi = df.groupby("year")["aqi"].mean()

print(yearly_aqi.round(2))
```

```python
highest_year = yearly_aqi.idxmax()
highest_year_aqi = yearly_aqi.max()

print("Year with highest average AQI:", highest_year)
print("Average AQI:", round(highest_year_aqi, 2))
```

```python
lowest_year = yearly_aqi.idxmin()
lowest_year_aqi = yearly_aqi.min()

print("Year with lowest average AQI:", lowest_year)
print("Average AQI:", round(lowest_year_aqi, 2))
```

```python
df["month"] = df["date"].dt.month
```

```python
monthly_aqi = df.groupby("month")["aqi"].mean()

print(monthly_aqi.round(2))
```

```python
top_10_aqi = df.nlargest(10, "aqi")

print(top_10_aqi[["city", "date", "aqi", "aqi_bucket"]])
```

```python
top_10_pm25 = df.nlargest(10, "pm2.5")

print(top_10_pm25[["city", "date", "pm2.5", "aqi"]])
```

```python
top_10_pm10 = df.nlargest(10, "pm10")

print(top_10_pm10[["city", "date", "pm10", "aqi"]])
```

```python
city_category = pd.crosstab(
    df["city"],
    df["aqi_bucket"]
)

print(city_category)
```

```python
print("========== AIR QUALITY ANALYSIS SUMMARY ==========")

print("\nTotal records:", len(df))
print("Total cities:", df["city"].nunique())

print("\nAverage AQI:", round(df["aqi"].mean(), 2))
print("Minimum AQI:", df["aqi"].min())
print("Maximum AQI:", df["aqi"].max())

print("\nHighest average AQI city:")
print(city_aqi.idxmax(), "->", round(city_aqi.max(), 2))

print("\nLowest average AQI city:")
print(city_aqi.idxmin(), "->", round(city_aqi.min(), 2))

print("\nMost common AQI category:")
print(df["aqi_bucket"].value_counts().idxmax())

print("\nHighest average PM2.5 city:")
print(pm25_city.idxmax(), "->", round(pm25_city.max(), 2))

print("\nHighest average PM10 city:")
print(pm10_city.idxmax(), "->", round(pm10_city.max(), 2))

print("\nYear with highest average AQI:")
print(yearly_aqi.idxmax(), "->", round(yearly_aqi.max(), 2))

print("\nYear with lowest average AQI:")
print(yearly_aqi.idxmin(), "->", round(yearly_aqi.min(), 2))
```
